```python
import matplotlib.pyplot as plt

for label in (tele.columns):
    plt.hist(tele[tele['class']=='g'][label],color="green",label="gamma",alpha=0.5,density=True)
    plt.hist(tele[tele['class']=='h'][label],color="yellow",label="hadron",alpha=0.5,density=True)

    plt.title(label)
    plt.ylabel("Probability")
    plt.xlabel(label)
    plt.legend()
    plt.show();
```

```python
from sklearn.model_selection import train_test_split

ds_train,ds_test=train_test_split(tele,test_size=0.2)
```

```python
print(len(ds_train))
print(len(ds_test))
```

```python
print(len(ds_train[ds_train['class']=='g']))
print(len(ds_train[ds_train['class']=='h']))
```

```python
from imblearn.over_sampling import RandomOverSampler

r_ram=RandomOverSampler()

in_ds_train=ds_train.drop(columns=['class'])
out_ds_train=ds_train['class']
```

```python
in_ds_train
```

```python
out_ds_train
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
in_ds_train=scaler.fit_transform(in_ds_train)
print(in_ds_train)
```

```python
in_ds_train,out_ds_train=r_ram.fit_resample(in_ds_train,out_ds_train)
print(len(out_ds_train))
print(sum(out_ds_train=='g'))
print(sum(out_ds_train=='h'))
```

```python
in_ds_test=ds_test.drop(columns=['class'])
out_ds_test=ds_test['class']
in_ds_test=scaler.fit_transform(in_ds_test)
```

```python
in_ds_test
```

```python
print(len(in_ds_test))
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(in_ds_train,out_ds_train)
```

```python
out_pred=knn.predict(in_ds_test)
```

```python
out_pred
```

```python
from sklearn.metrics import accuracy_score

accuracy_info=accuracy_score(out_ds_test,out_pred)
```
