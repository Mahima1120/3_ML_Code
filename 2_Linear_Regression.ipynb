```python
ip=car.drop(columns=["car_name","selling_price"])
op=car["selling_price"]
```

```python
ip
```

```python
op
```

```python
print(ip.shape)
print(op.shape)
```

```python
import pandas as pd

ip=pd.get_dummies(ip,dtype=int)
print(ip.head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
ip=scaler.fit_transform(ip)
print(ip)
```

```python
from sklearn.model_selection import train_test_split

ip_train,ip_test,op_train,op_test=train_test_split(ip,op,test_size=0.2)
```

```python
print(ip_train.shape)
```

```python
print(ip_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn import metrics

model=LinearRegression()
model.fit(ip_train,op_train)
```

```python
predict=model.predict(ip_test)
```

```python
predict
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
```

```python
print("mean absolute error",(metrics.mean_absolute_error(predict,op_test)))
print("mean square error",(metrics.mean_squared_error(predict,op_test)))
```
